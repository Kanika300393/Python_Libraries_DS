# Pandas for Data Visualisation

```python
#importing libraries
import numpy as np
import pandas as pd 
%matplotlib inline
```

```python
dates = pd.date_range('1/1/2000', periods = 500 ) #generating 500 dates randomly
```

```python
col_D = np.random.rand(500) #generating column col_D with anoth 500 numbers
```

```python
data1 = np.random.rand(500,3)
```

```python
df1 = pd.DataFrame(data = data1, index = dates , columns = ['A','B','C'])
```

```python
df1.head()
```

```python
df1['D'] = col_D
```

```python
df1.head()
```

```python
data2 = np.random.rand(20,3)
col = ['X','Y','Z']
df2 = pd.DataFrame(data = data2, columns=col)
```

```python
df2.head()
```

```python
#Histogram plot - two ways
df1['A'].hist()
```

```python
df1['A'].hist(bins=50)
```

```python
#Another way of plotting histogram
df1['A'].plot(kind = 'hist')
```

```python
df1['A'].plot(kind = 'hist', bins = 50)
```

The only difference between two graphs in second one we are getting the white background, but in 1st we are getting the grids 

### Lets work with style sheets

### Scatter Plot

```python
df1.plot.scatter(x='A', y= 'B') #It is recomended to use different color 
```

```python
ax = df1.plot.scatter(x = 'A', y = 'B', color = 'DarkBlue') 
# We have the same scatter plot in darkblue color 
```

```python
ax = df1.plot.scatter(x = 'A', y = 'B', color = 'DarkBlue', label = 'Group1')
df1.plot.scatter(x = 'C', y = 'D', color = 'DarkGreen', ax= ax, label = 'Group2') 
```

```python
df1.plot.scatter(x='A', y ='B', c = 'C', cmap= 'coolwarm')
```

```python
df1.plot.scatter(x='A', y ='B', c = 'C', cmap= 'coolwarm', s = 60) #INCREASING THE SIZE OF DATA POINTS 
```

```python
df1.plot.scatter(x='A', y ='B',  cmap= 'coolwarm', s = df1['C']*100) 
# Size of the bubbles depend upon size of the value of  C
```
