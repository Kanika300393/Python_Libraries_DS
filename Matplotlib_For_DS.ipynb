### Matplotlib Object oriented Approach 

```python
fig1 = plt.figure()
axes = fig1.add_axes([0.1,0.1,0.9,0.9])
axes.plot(x,y)
```

We have seen the basic plotting which is very easy and quick to generate the plots. However, It is recommended to use object oriented approach for more control and customisation of our plots.

The idea behind Object oriented approach is create a figure object and then call methods and attributes from that object.
This approach is helpful when we dealing with canvas that has multiple plots on it.

```python
#Lets create a figure instance say a fig1 and add axes on that figure
fig1 = plt.figure() # create fig1 as an object 
axes = fig1.add_axes([0.1,0.1,0.9,0.9]) # (0.1 = Length,0.1 = bottom,0.9= width,0.9= height )
axes.plot(x,y)
axes.set_xlabel('X')
axes.set_ylabel('Y')
axes.set_title('Title')
```

```python
#we are going to create a canva here and inside that graph , we are going to plot another graph

fig2 = plt.figure()
axes_main = fig2.add_axes([0.1,0.1,0.9,0.9])
axes_inset = fig2.add_axes([0.3,0.5,0.3,0.4])
axes_main.plot(x,y,'b')
axes_inset.plot(y,x,'g')

axes_main.set_xlabel('Main X')
axes_main.set_ylabel('Main Y')
axes_main.set_title('Main Title')

axes_inset.set_xlabel('Inset X')
axes_inset.set_ylabel('Inset Y')
axes_inset.set_title('Inset Title')
```

### Lets Talk about Figure Size, aspect ratio and DPI 

While creating a figure object, Matplotlib allows aspect ratio, Dpi and figure size to be specify.
Fig size is actually the weight than the height in the inches.
DPI is Dot Spot Per Inch or Pixel per inch  

```python
fig = plt.figure(figsize=(10,5),dpi=100)
```

## How to decorate the image

```python
fig = plt.figure()
axes = fig.add_axes([0,0,1,1])
axes.plot(x, x**2, label = 'x**2')
axes.plot(x, x**3, label = 'x**3')
axes.legend(loc = 2) #simple way to call the legend 
# loc = 1 , top right corner
# loc = 2, top left corner
# loc = 3, lower left corner
# loc = 4, lower right corner
# loc = 0 # then matplotlib decide where to plot the legends
```

```python
fig, axes = plt.subplots()
axes.plot(x,x+1,'r.-')
axes.plot(x,x+2,'g--')
```

```python
fig, axes = plt.subplots()
axes.plot(x,x+1,color = 'blue', alpha = 0.5) # HERE alpha means transperency of line
axes.plot(x,x+2,color = '#8B008B')
axes.plot(x,x+3,color = '#FF8C00')
```

```python
#Checking the same code with alpha = 0.1, see its very transparent
fig, axes = plt.subplots()
axes.plot(x,x+1,color = 'blue', alpha = 0.1) # HERE alpha means transperency of line
axes.plot(x,x+2,color = '#8B008B')
axes.plot(x,x+3,color = '#FF8C00')
```

```python
fig, axes = plt.subplots(figsize = (10,8))
axes.plot(x,y, color = 'purple', lw = 4, ls = '-' )# lw = line width, ls = Line style
```

```python
#passing the markers as well
fig, axes = plt.subplots(figsize = (10,8))
axes.plot(x,y, color = 'purple', lw = 4, ls = '-', marker = 's', markersize = 18 ,
         markerfacecolor= 'yellow', markeredgewidth= 4,markeredgecolor ='red' )# lw = line width, ls = Line style
```

```python
fig, axes = plt.subplots(nrows = 1, ncols = 3, figsize=(10,3))
axes[0].plot(x,y)
axes[0].set_title('default')

axes[1].plot(x,y)
axes[1].axis('tight')
axes[1].set_title('tight')

axes[2].plot(x,y)
axes[2].set_title('custom')
axes[2].set_xlim([1,4])
axes[2].set_ylim([0,50])
```

np.random.normal(0, std, 100):

This generates 100 random numbers from a normal distribution (also known as a Gaussian distribution).
The mean (loc) of the distribution is 0.
The standard deviation (scale) of the distribution is std, which changes with each iteration of the loop.
100 is the number of random values generated for each iteration.
for std in range(1, 4):

This loop iterates over the values 1, 2, and 3.
In each iteration, std takes one of these values, meaning the standard deviation 
used in the normal distribution changes from 1 to 2 to 3.

The list will contain three arrays (one for each standard deviation: 1, 2, and 3).
