# Seaborn - Statistical Data Visualisation Library 

Seaborn is a powerful and user-friendly data visualization library in Python, built on top of Matplotlib. It provides a high-level interface for creating attractive and informative statistical graphics. Seaborn is designed to make it easier to create complex visualizations with just a few lines of code, and it integrates well with Pandas, allowing you to easily plot data from data frames.

### Key Features:

##### 1. Enhanced Aesthetics: 
Seaborn offers a variety of built-in themes and color palettes to make your plots look visually appealing by default.

##### 2. Statistical Plots: 
Seaborn simplifies the creation of common statistical plots like bar plots, box plots, violin plots, pair plots, and heatmaps, among others.

##### 3. Complex Plots Made Easy:
It allows you to create complex visualizations, such as multi-plot grids and facet plots, with simple commands.

##### 4. Integration with Pandas:
Seaborn works seamlessly with Pandas DataFrames, making it easy to plot data directly from your data structures.

##### 5. Built-in Statistical Functions: 
Seaborn includes functions to add linear regression lines, compute means, medians, confidence intervals, etc., directly within your plots.

Seaborn comes with built in datasets that can be loaded using the functions load_dataset('datasetname')

### Part 1 (Distribution Plots)

Distribution plots include: distplot(), jointplot(), pairplot(), rugplot(), kdeplot() etc

```python
import seaborn as sns
%matplotlib inline
```

```python
sns.get_dataset_names()
```

```python
tips = sns.load_dataset('tips')
```

```python
tips.info
```

```python
tips.head()
```

```python
#lets start with displot, to check the univariate distribution, distplot is a good choice.

sns.distplot(tips['total_bill']) #checking distribution of the variable total_bill from distplot
#here line over the graph is KDE
```

```python
sns.distplot(tips['total_bill'], kde = False, bins = 50) #run the same code with kde false , so we can not have the line
```

#### So the above graph shows that most of the bills are between 10 and 20 

```python
#Joint plot
# We want to do the comparison
#like how much tips are paid by customers based on the bills

sns.jointplot(x = 'total_bill', y ='tip', data = tips)

#default is scatter plot
#this is just 2 distribution plot 
```

#### Most of the tips are less than 4 dollars

#### More the customer is paying the bill, more customer is paying the tip

```python
sns.jointplot(x = 'total_bill', y ='tip', data = tips, kind = 'hex') #now its different
#here darker the color means, more datapoints are there, lesser the color means less data points are there
```

```python
sns.jointplot(x = 'total_bill', y ='tip', data = tips, kind = 'reg') #reg = linear regression
```

```python
sns.jointplot(x = 'total_bill', y ='tip', data = tips, kind = 'kde') #This is 2D density plot.
```

```python
sns.pairplot(tips)
# For larger dataset it will take time to plot the pair plot 
```

```python
sns.pairplot(tips, hue = 'sex') #here we have pair plot in two different colors, we passed the sex because it is categorical
```

```python
sns.rugplot(tips['total_bill'])
```

```python
#rugplot Vs Distplot
sns.rugplot(tips['total_bill'], color = 'r')
sns.distplot(tips['total_bill'], kde = False)

#The more lines in the rugplot, the higher we have the displot
```

```python
sns.kdeplot(tips['total_bill']) #Line in the graph
sns.rugplot(tips['total_bill']) #bins in the graph
sns.distplot(tips['total_bill']) #orange lines at the bottom stacked
```

```python
sns.kdeplot(tips['tip']) #Line in the graph
sns.rugplot(tips['tip']) #bins in the graph
sns.distplot(tips['tip']) #orange lines at the bottom stacked
```

### Part 2 - Categorical Data

Categorical data is converted into three plots.:- 

Swarn Plot, Strip Plot :- are those that shows each observation at each level of the categorical variable.

boxplot(), violinplot():- are those that show an abstract representation of each distribution of observation.

barplot(), pointplot() :- are those that apply a statistical estimation to show a measure of
central tendency and confidence interval.  

```python
tips = sns.load_dataset('tips') #again loading the tips dataset
```

```python
tips.head() #checking the top 5 rows of the dataset 
```

### Strip Plot

```python
sns.stripplot(x= 'day', y ='total_bill', data = tips, jitter = False)
```

Here we see that data points are overlapped, this is common but makes difficult to understand distribution in our data.
There is a parameter "Jitter" used with this code which resolve this problem

```python
sns.stripplot(x= 'day', y ='total_bill', data = tips, jitter = True )
```

Now, we can see that datapoints are not overlapping 

```python
sns.stripplot(x= 'day', y ='total_bill', data = tips, jitter = True , palette ='coolwarm', hue ='sex')
```

Now, Blue are males and orange are Females

```python
sns.stripplot(x= 'day', y ='total_bill', data = tips, jitter = True , hue ='sex')
```

Pallete was optional, if we are going without that , It will give you by default dark pic

There are several types of axis grids in seaborn:
FacetGrid: 
PairGrid:
JointGrid:
etc

When we are exploring a medium - dimensional data, a useful approach is to draw multiple instances of the same plot on different subsets of our dataset. Grid Chart/ plots (also called panel , lattice or trellis charts/plots) aloows us to quickly extract significant amount of information about the complex data in a series of similar graphs using the same scale and axes. This allows easily comparisons and provides a brief  overview of the data under analysis.

## Matrix Plot

In the Seaborn library, matrix plots are used to visualize relationships between variables in a dataset, often represented in matrix form. Here are the common types of matrix plots in Seaborn:

Heatmap (sns.heatmap)

Clustermap (sns.clustermap)

Pairplot (sns.pairplot)


Heatmap allows us to plot  the data as color encoded matrices, whereas
clustermap can be used to indicate clusters within the data

```python
tips = sns.load_dataset('tips')
flights = sns.load_dataset('flights')
```

```python
flights.info()
```

```python
flights.head()
```

```python
tips.head()
```

```python
print(tips.dtypes)
```

Here, as we have both categorical and numerical data so wither we can convert the categorical data into numerical, or direct use the number data for the coorelation matrix

```python
tips_numerical = tips.select_dtypes(include='number') 
correlation_matrix = tips_numerical.corr()
print(correlation_matrix)
```

```python
sns.heatmap(correlation_matrix)
```

```python
sns.heatmap(correlation_matrix, cmap='rainbow' ) #Another parameter cmap
```

```python
sns.heatmap(correlation_matrix, cmap='coolwarm' ) #Another variable
```

```python
sns.heatmap(correlation_matrix, cmap='coolwarm' , annot = True) #Passing another variable annote which displays valeues
```

The sns.despine() function in Seaborn is used to remove or "despine" the top and right spines (the borders) of a plot. This function is useful for creating cleaner, more minimalist visualizations by reducing unnecessary chart elements, making the data stand out more clearly.

```python
sns.countplot(x = 'sex', data = tips)
sns.despine(top = False, right = False, left = False, bottom = False)
```

```python
sns.countplot(x = 'sex', data = tips)
sns.despine(top = True, right = True, left = True, bottom = True)

# Now we dont have any spine in the graph
```
